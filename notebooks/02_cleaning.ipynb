```python
import pandas as pd
from pathlib import Path
```

```python
RAW_DATA_PATH = Path("../data/raw")
CLEAN_DATA_PATH = Path("../data/cleaned")

CLEAN_DATA_PATH.mkdir(parents=True, exist_ok=True) 
```

```python
df = pd.read_csv(
    RAW_DATA_PATH /
    "BTCUSD_2010-01-01_2025-12-31_raw.csv")

df.head()
```

```python
print(df.shape)

df.info()
```

```python
df["date"] = pd.to_datetime(df["date"]) #conversion des dates
```

```python
df = ( df.sort_values("date").reset_index(drop=True))

# Nous sommes en serie temporelle. Donc l'ordre d'observation 
# est critique. 
```

```python
print( "Doublons :", df.duplicated().sum())
```

```python
df = df.drop_duplicates() #suppression des doublons
```

```python
missing = (df.isna().sum().sort_values(ascending=False))

print(missing)

# affiichage des valeurs manquantes
```

```python
numeric_cols = ["open","high","low","close","volume"]

for col in numeric_cols:

    df[col] = pd.to_numeric(df[col],errors="coerce")
```

```python
df.to_csv( CLEAN_DATA_PATH / "BTCUSD_clean.csv", index=False )

print("Dataset nettoyé sauvegardé.")
```
