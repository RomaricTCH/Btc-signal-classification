```python
import requests
import pandas as pd
from pathlib import Path
```

```python
# Paramètres

import os
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.getenv("FMP_API_KEY")

SYMBOL = "BTCUSD"

START_DATE = "2010-01-01"
END_DATE = "2025-12-31"

RAW_DATA_PATH = Path("../data/raw")

RAW_DATA_PATH.mkdir(parents=True, exist_ok=True )
```

```python
print(API_KEY is not None)
```

```python
def fetch_fmp_crypto_data(symbol, start_date, end_date, api_key):

    url = ( f"https://financialmodelingprep.com/stable/historical-price-eod/full"
            f"?symbol={symbol}&from={start_date}&to={end_date}&apikey={api_key}")


    response = requests.get(url)

    if response.status_code != 200:
        raise Exception(f"Erreur API : {response.status_code}\n"
                        f"{response.text}")

    data = response.json()
    df = pd.DataFrame(data)

    if df.empty:
        raise ValueError("Aucune donnée récupérée depuis l'API.")

    df["date"] = pd.to_datetime(df["date"])

    df = df[
        (df["date"] >= pd.to_datetime(start_date)) & (df["date"] <= pd.to_datetime(end_date))]

    return df
```

```python
df_raw = fetch_fmp_crypto_data(
    symbol=SYMBOL,
    start_date=START_DATE,
    end_date=END_DATE,
    api_key=API_KEY )

df_raw.head()
```

```python
print(df_raw["date"].min())
print(df_raw["date"].max())

print("\nNombre de lignes :", len(df_raw))
```

```python
df_raw.to_csv(
    RAW_DATA_PATH / f"{SYMBOL}_{START_DATE}_{END_DATE}_raw.csv",
    index=False
)

print("Données sauvegardées.")
```
